```python
import cv2
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = [4, 3]
img=cv2.imread('gato.jpg', cv2.IMREAD_COLOR)
#imagen original y dividida por canales

plt.subplot(141)
plt.title(' imagen Original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title('Canal Rojo')
plt.imshow(img[:,:,0], cmap='gray')

plt.subplot(143)
plt.title('Canal Verde')
plt.imshow(img[:,:,1], cmap='gray')

plt.subplot(144)
plt.title('Canal Azul')
plt.imshow(img[:,:,2], cmap='gray')
plt.show()
```

con colores rellenos

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [10, 5]
img=cv2.imread('gato.jpg',cv2.IMREAD_COLOR)
imgRGB=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r=imgRGB[:,:,2]
g=imgRGB[:,:,1]
b=imgRGB[:,:,0]

#Crear imagen por canal
r_img=np.zeros_like(imgRGB)
r_img[:,:,0]=r

g_img=np.zeros_like(imgRGB)
g_img[:,:,1]=g

b_img=np.zeros_like(imgRGB)
b_img[:,:,2]=b

plt.figure(figsize=(10,5))
plt.subplot(141)
plt.title("IMAGEN ORIGINAL")
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(142)
plt.title("CANAL ROJO")
plt.imshow(r_img)
plt.axis('off')

plt.subplot(143)
plt.title("CANAL VERDE")
plt.imshow(g_img)
plt.axis('off')

plt.subplot(144)
plt.title("CANAL AZUL")
plt.imshow(b_img)
plt.axis('off') 


plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [20, 8]
img=cv2.imread('gato.jpg',cv2.IMREAD_COLOR)
imgRGB=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

#Primera fila
plt.subplot(2,6,1)
plt.title("IMAGEN ORIGINAL")
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(2,6,2)
plt.title("FALSO COLOR GRAY")
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.subplot(2,6,3)
plt.title("FALSO COLOR VERDE")
plt.imshow(gray, cmap='viridis')
plt.axis('off')

plt.subplot(2,6,4)
plt.title("FALSO COLOR PLASMA")
plt.imshow(gray, cmap='plasma')
plt.axis('off')

plt.subplot(2,6,5)
plt.title("FALSO COLOR INFERNO")
plt.imshow(gray, cmap='inferno')
plt.axis('off')

plt.subplot(2,6,6)
plt.title("FALSO COLOR MAGMA")
plt.imshow(gray, cmap='magma')
plt.axis('off')

#Segunda fila
plt.subplot(2,6,7)
plt.title("FALSO COLOR CASI NORMAL")
plt.imshow(gray, cmap='cividis')
plt.axis('off')

plt.subplot(2,6,8)
plt.title("FALSO COLOR HOTCITO")
plt.imshow(gray, cmap='hot')
plt.axis('off')

plt.subplot(2,6,9)
plt.title("FALSO COLOR CELESTE")
plt.imshow(gray, cmap='cool')
plt.axis('off')

plt.subplot(2,6,10)
plt.title("FALSO COLOR ROSA")
plt.imshow(gray, cmap='spring')
plt.axis('off')

plt.subplot(2,6,11)
plt.title("FALSO COLOR rojo vivo")
plt.imshow(gray, cmap='autumn')
plt.axis('off')

plt.show()
```

imagenes

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
img= cv2.imread('gato.jpg')
img_rgb=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].imshow(img_rgb)
axs[0].set_title('Imagen Original')

fig_img=axs[1].imshow(img_rgb[...,0], cmap='gray')
fig.colorbar(fig_img, ax=axs[1],shrink=0.5)
axs[1].set_title('Canal Rojo')

fig_img=axs[2].imshow(img_rgb[...,1], cmap='gray')
fig.colorbar(fig_img, ax=axs[2],shrink=0.5)
axs[2].set_title('Canal Verde')

fig_img=axs[3].imshow(img_rgb[...,2], cmap='gray')
fig.colorbar(fig_img, ax=axs[3],shrink=0.5)
axs[3].set_title('Canal Azul1')
```

EJERCICIOS PARA GENERAR UNA IMAGEN

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [10, 5]
size_imagen=100
R=np.ones((size_imagen,size_imagen))*255
G=np.ones((size_imagen,size_imagen))*128
B=np.ones((size_imagen,size_imagen))*0

imagen=np.dstack((R,G,B))
imagen=np.array(imagen,dtype=int)

plt.figure()
plt.imshow(imagen)
plt.title('imagen color:naranja')
plt.show
```
