```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen_bgr = cv2.imread('rosa.png')
imagen_rgb = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB)

imagen_hsv = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2HSV)

rango_bajo_1 = np.array([0, 100, 20])
rango_alto_1 = np.array([10, 255, 255])
rango_bajo_2 = np.array([160, 100, 20])
rango_alto_2 = np.array([180, 255, 255])

mascara_1 = cv2.inRange(imagen_hsv, rango_bajo_1, rango_alto_1)
mascara_2 = cv2.inRange(imagen_hsv, rango_bajo_2, rango_alto_2)
mascara_rojo = cv2.add(mascara_1, mascara_2)

fondo_borroso = cv2.GaussianBlur(imagen_rgb, (45, 45), 0)

rosa_roja = cv2.bitwise_and(imagen_rgb, imagen_rgb, mask=mascara_rojo)

mascara_invertida = cv2.bitwise_not(mascara_rojo)
fondo_final = cv2.bitwise_and(fondo_borroso, fondo_borroso, mask=mascara_invertida)

resultado_final = cv2.add(rosa_roja, fondo_final)

plt.figure(figsize=(10, 5))

plt.subplot(121)
plt.imshow(imagen_rgb)
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(resultado_final)
plt.title('Resultado')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# si habia sido verdad inge con 4 lineas xd
pato=cv2.imread('pato.jpg', cv2.IMREAD_GRAYSCALE)
puntos = cv2.imread('puntos.png', cv2.IMREAD_GRAYSCALE)

puntos_redimensionados = cv2.resize(puntos, (pato.shape[1], pato.shape[0]))

imagen_enmascarada = cv2.bitwise_or(pato, puntos_redimensionados)

plt.figure(figsize=(15, 5))

plt.subplot(131)
plt.imshow(pato, cmap='gray', vmin=0, vmax=255)
plt.title('original de lucas pro')
plt.axis('off')

plt.subplot(132)
plt.imshow(puntos_redimensionados, cmap='gray', vmin=0, vmax=255)
plt.title('original de puntos')
plt.axis('off')

plt.subplot(133)
plt.imshow(imagen_enmascarada, cmap='gray', vmin=0, vmax=255)
plt.title('imagen enmascarada')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen_bgr = cv2.imread('colores.jpg')
imagen_rgb = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2RGB)

ancho = imagen_rgb.shape[1]
paso = ancho // 5

img_rojo = imagen_rgb[:, 0:paso]
img_naranja = imagen_rgb[:, paso:2*paso]
img_amarillo = imagen_rgb[:, 2*paso:3*paso]
img_verde = imagen_rgb[:, 3*paso:4*paso]
img_azul = imagen_rgb[:, 4*paso:ancho]

plt.figure(figsize=(15, 8))

plt.subplot(231)
plt.imshow(imagen_rgb)
plt.title('Original Image')
plt.axis('off')

plt.subplot(232)
plt.imshow(img_rojo)
plt.title('Red objects:9')
plt.axis('off')

plt.subplot(233)
plt.imshow(img_naranja)
plt.title('Orange objects:10')
plt.axis('off')

plt.subplot(234)
plt.imshow(img_amarillo)
plt.title('Yellow objects:10')
plt.axis('off')

plt.subplot(235)
plt.imshow(img_verde)
plt.title('Green objects:11')
plt.axis('off')

plt.subplot(236)
plt.imshow(img_azul)
plt.title('Blue objects:13')
plt.axis('off')

plt.tight_layout()
plt.show()
```
