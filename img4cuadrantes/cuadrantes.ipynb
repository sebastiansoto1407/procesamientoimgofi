```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [6, 6]
size_imagen = 200
imagen = np.zeros((size_imagen, size_imagen, 3), dtype=np.uint8)
colores = [[255, 128, 0], [255, 5, 0], [0, 255, 0], [0, 103, 226]]
mitad = size_imagen // 2
k = 0
for f in range(2):
    for c in range(2):
        y0 = f * mitad
        y1 = (f + 1) * mitad
        x0 = c * mitad
        x1 = (c + 1) * mitad
        imagen[y0:y1, x0:x1] = colores[k]
        k += 1
plt.figure()
plt.imshow(imagen)
plt.title('4 cuadraditos RGB')
plt.axis('off')
plt.show()


```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [6, 6]
size_cuadro = 100
size_imagen = size_cuadro * 3
imagen = np.zeros((size_imagen, size_imagen, 3), dtype=np.uint8)
colores_rubik = np.array([[255, 0, 0], [0, 255, 0], [0, 0, 255], [255, 255, 255], [255, 128, 0], [255, 255, 0]])
colores_random = np.random.choice(len(colores_rubik), 9)
k = 0
for f in range(3):
    for c in range(3):
        y0 = f * size_cuadro
        y1 = (f + 1) * size_cuadro
        x0 = c * size_cuadro
        x1 = (c + 1) * size_cuadro
        imagen[y0:y1, x0:x1] = colores_rubik[colores_random[k]]
        k += 1
plt.figure()
plt.imshow(imagen)
plt.title('Rubik 3x3 random')
plt.axis('off')
plt.show()
```
