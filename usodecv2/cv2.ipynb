Cargar 3 im3genes, convertir a RGB y gris, y mostrar en diferentes modos...

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

#funcion
def mostrar_imagen(imagen, titulo, es_gris=False):
    plt.figure(figsize=(6, 4))
    if es_gris:
        plt.imshow(imagen, cmap='gray')
    else:
        plt.imshow(imagen)
    plt.title(titulo)
    plt.axis('off')
    plt.show()
```

## Foto 1

```python
#cargar y guardar foto 1
img = cv2.imread('cv1.jpeg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imwrite('cv1_bgr.png', img)
cv2.imwrite('cv1_gray.png', gray)

mostrar_imagen(img_rgb, 'Foto 1 - original')

mostrar_imagen(gray, 'Foto 1 - escala de grises', es_gris=True)
```

```python
#foto 1 canales en gris
plt.rcParams['figure.figsize'] = [4, 3]
img = cv2.imread('cv1.jpeg')

plt.subplot(141)
plt.title('original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title('canal rojo')
plt.imshow(img[:,:,0], cmap='gray')

plt.subplot(143)
plt.title('canal verde')
plt.imshow(img[:,:,1], cmap='gray')

plt.subplot(144)
plt.title('canal azul')
plt.imshow(img[:,:,2], cmap='gray')
plt.show()
```

```python
#foto 1 canales en color
img = cv2.imread('cv1.jpeg')
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = imgRGB[:,:,2]
g = imgRGB[:,:,1]
b = imgRGB[:,:,0]

r_img = np.zeros_like(imgRGB)
r_img[:,:,0] = r

g_img = np.zeros_like(imgRGB)
g_img[:,:,1] = g

b_img = np.zeros_like(imgRGB)
b_img[:,:,2] = b

plt.figure(figsize=(10, 5))
plt.subplot(141)
plt.title('original')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(142)
plt.title('canal rojo')
plt.imshow(r_img)
plt.axis('off')

plt.subplot(143)
plt.title('canal verde')
plt.imshow(g_img)
plt.axis('off')

plt.subplot(144)
plt.title('canal azul')
plt.imshow(b_img)
plt.axis('off')
plt.show()
```

```python
#foto 1 falso color
plt.rcParams['figure.figsize'] = [20, 8]
img = cv2.imread('cv1.jpeg')
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.subplot(2,6,1)
plt.title('original')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(2,6,2)
plt.title('gray')
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.subplot(2,6,3)
plt.title('viridis')
plt.imshow(gray, cmap='viridis')
plt.axis('off')

plt.subplot(2,6,4)
plt.title('plasma')
plt.imshow(gray, cmap='plasma')
plt.axis('off')

plt.subplot(2,6,5)
plt.title('inferno')
plt.imshow(gray, cmap='inferno')
plt.axis('off')

plt.subplot(2,6,6)
plt.title('magma')
plt.imshow(gray, cmap='magma')
plt.axis('off')

plt.subplot(2,6,7)
plt.title('cividis')
plt.imshow(gray, cmap='cividis')
plt.axis('off')

plt.subplot(2,6,8)
plt.title('hot')
plt.imshow(gray, cmap='hot')
plt.axis('off')

plt.subplot(2,6,9)
plt.title('cool')
plt.imshow(gray, cmap='cool')
plt.axis('off')

plt.subplot(2,6,10)
plt.title('spring')
plt.imshow(gray, cmap='spring')
plt.axis('off')

plt.subplot(2,6,11)
plt.title('autumn')
plt.imshow(gray, cmap='autumn')
plt.axis('off')

plt.show()
```
