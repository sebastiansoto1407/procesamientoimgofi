```python
# ejer2

n = int(input("Escribe un número entero: "))

es_primo = True

if n <= 1:
    es_primo = False
else:
    divisor = 2
    while divisor < n:
        if n % divisor == 0:
            es_primo = False
            break
        divisor = divisor + 1

if es_primo:
    print("El numero", n, "es primo")
else:
    print("El numero", n, "NO es primo")
```

```python
#ejer3

nums = []
limite = 3
i = 0

print("Escribe hasta 3 numeros (3 veces tiene q escoger):")

while i < limite:
    entrada = input("Número: ")
    if entrada == "":
        break
    try:
        num = int(entrada)
        nums.append(num)
        i += 1
    except ValueError:
        print("Escribe solo numeros enteros")

if nums:
    mayor = nums[0]
    menor = nums[0]
    
    for num in nums:
        if num > mayor:
            mayor = num
        if num < menor:
            menor = num
    
    print("Mayor:", mayor)
    print("Menor:", menor)
else:
    print("No se ingresaron numeros")
```

```python
#ejer 4

texto = input("Escribe una cadena: ").lower()
vocales = "aeiou"
contador_vocales = 0

for letra in texto:
    if letra in vocales:
        contador_vocales += 1

print("Vocales:", contador_vocales)
```

```python
#ejer5

texto = input("Escribe cadena: ").lower().replace(" ", "")
es_palindromo = True
inicio = 0
fin = len(texto) - 1

while inicio < fin:
    if texto[inicio] != texto[fin]:
        es_palindromo = False
        break
    inicio += 1
    fin -= 1

print("Es palindromo:", es_palindromo)
```

```python
#ejer7

nums = []
n = int(input("Cantidad de numeros: "))
for i in range(n):
    num = int(input("Numero: "))
    nums.append(num)

for i in range(n-1):
    for j in range(n-i-1):
        if nums[j] > nums[j+1]:
            nums[j], nums[j+1] = nums[j+1], nums[j]

print("Ordenados:", nums)
```
