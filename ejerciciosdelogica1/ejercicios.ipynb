```python
# ejer2

n = int(input("Escribe un número entero: "))

es_primo = True

if n <= 1:
    es_primo = False
else:
    divisor = 2
    while divisor < n:
        if n % divisor == 0:
            es_primo = False
            break
        divisor = divisor + 1

if es_primo:
    print("El numero", n, "es primo")
else:
    print("El numero", n, "NO es primo")
```
