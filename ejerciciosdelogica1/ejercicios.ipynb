```python
# ejercicio 1 :=) y prueba si funciona
numero_entero_ingresado_por_el_usuario = int(
    input("Por favor, escribe un num entero para calcular su factorial: ")
)

resultado_factorial_del_numero_ingresado = 1

if numero_entero_ingresado_por_el_usuario < 0:
    print("El factorial de un numero negativo no esta definido")
else:
    
    numero_actual_que_se_esta_multiplicando = 1

    while numero_actual_que_se_esta_multiplicando <= numero_entero_ingresado_por_el_usuario:
        resultado_factorial_del_numero_ingresado = (
            resultado_factorial_del_numero_ingresado * numero_actual_que_se_esta_multiplicando
        )
        numero_actual_que_se_esta_multiplicando = (
            numero_actual_que_se_esta_multiplicando + 1
        )

    print(
        "El factorial de",
        numero_entero_ingresado_por_el_usuario,
        "es",
        resultado_factorial_del_numero_ingresado,
    )
```
