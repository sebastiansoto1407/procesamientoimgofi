```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

img_path = Path("laplacianofoto.jpeg")
img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)

#laplaciano directo
lap = cv2.Laplacian(img, cv2.CV_64F, ksize=3)
lap_abs = cv2.convertScaleAbs(lap)

#LoG: suavizar con Gaussiano y luego Laplaciano
img_blur = cv2.GaussianBlur(img, (5, 5), sigmaX=1.2)
log = cv2.Laplacian(img_blur, cv2.CV_64F, ksize=3)
log_abs = cv2.convertScaleAbs(log)

plt.figure(figsize=(12, 4))
titles = ["Original", "Laplaciano", "LoG (Gauss + Lap)"]
images = [img, lap_abs, log_abs]
for i, (t, im) in enumerate(zip(titles, images), 1):
    plt.subplot(1, 3, i)
    plt.title(t)
    plt.imshow(im, cmap="gray")
    plt.axis("off")
plt.tight_layout()
plt.show()
```
