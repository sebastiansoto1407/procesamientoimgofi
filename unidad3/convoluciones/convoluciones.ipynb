```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img_color=cv2.imread('soldierboy.jpg')
img_color=cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
img_gray=cv2.cvtColor(img_color, cv2.COLOR_RGB2GRAY)

kernels=[
    ("enfoque", np.array([(0,-1,0),(-1,5,-1),(0,-1,0)], dtype=np.float32)),
    ("desenfoque", np.array([(1,1,1),(1,1,1),(1,1,1)], dtype=np.float32)/9.0),
    ("realce de bordes", np.array([(0,0,0),(-1,1,0),(0,0,0)], dtype=np.float32)),
    ("repujado", np.array([(-2,-1,0),(-1,1,1),(0,1,2)], dtype=np.float32)),
    ("deteccion de bordes", np.array([(0,1,0),(1,-4,1),(0,1,0)], dtype=np.float32)),
    ("filtro de tipo sobel", np.array([(-1,0,1),(-2,0,2),(-1,0,1)], dtype=np.float32)),
    ("filtro de tipo sharpen", np.array([(1,-2,1),(-2,5,-2),(1,-2,1)], dtype=np.float32)),
    ("filtro norte", np.array([(1,1,1),(1,-2,1),(-1,-1,-1)], dtype=np.float32)),
    ("filtro este", np.array([(-1,1,1),(-1,-2,1),(-1,1,1)], dtype=np.float32)),
    ("filtro de tipo gauss", np.array([(1,2,3,1,1),(2,7,11,7,2),(3,11,17,11,3),(2,7,11,7,1),(1,2,3,2,1)], dtype=np.float32)/119.0)
 ]

def mostrar_resultados(imagen, es_gris=False):
    plt.figure(figsize=(18,6))
    plt.subplot(2,6,1)
    if es_gris:
        plt.imshow(imagen, cmap='gray')
    else:
        plt.imshow(imagen)
    plt.title('original')
    plt.axis('off')

    for i, (nombre, kernel) in enumerate(kernels, start=2):
        salida=cv2.filter2D(imagen, -1, kernel)
        plt.subplot(2,6,i)
        if es_gris:
            plt.imshow(salida, cmap='gray')
        else:
            plt.imshow(salida)
        plt.title(nombre)
        plt.axis('off')

    plt.tight_layout()
    plt.show()

mostrar_resultados(img_gray, es_gris=True)
mostrar_resultados(img_color, es_gris=False)
```
