 Practica: Paletas de colores en OpenCV

 cv2
 numpy
 matplotlib

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen_bgr = cv2.imread('../clase2/kermit.jpg')
imagen_gray = cv2.cvtColor(imagen_bgr, cv2.COLOR_BGR2GRAY)
 
paletas = [
    (0, 'COLORMAP_AUTUMN', cv2.COLORMAP_AUTUMN),
    (1, 'COLORMAP_BONE', cv2.COLORMAP_BONE),
    (2, 'COLORMAP_JET', cv2.COLORMAP_JET),
    (3, 'COLORMAP_WINTER', cv2.COLORMAP_WINTER),
    (4, 'COLORMAP_RAINBOW', cv2.COLORMAP_RAINBOW),
    (5, 'COLORMAP_OCEAN', cv2.COLORMAP_OCEAN),
    (6, 'COLORMAP_SUMMER', cv2.COLORMAP_SUMMER),
    (7, 'COLORMAP_SPRING', cv2.COLORMAP_SPRING),
    (8, 'COLORMAP_COOL', cv2.COLORMAP_COOL),
    (9, 'COLORMAP_HSV', cv2.COLORMAP_HSV),
    (10, 'COLORMAP_PINK', cv2.COLORMAP_PINK),
    (11, 'COLORMAP_HOT', cv2.COLORMAP_HOT),
]

# Visualizacion con ciclo repetitivo
filas = len(paletas)
fig, axes = plt.subplots(filas, 1, figsize=(7, 2.0 * filas))

if filas == 1:
    axes = [axes]

for i, (valor, nombre, codigo) in enumerate(paletas):
    imagen_color = cv2.applyColorMap(imagen_gray, codigo)
    imagen_color = cv2.cvtColor(imagen_color, cv2.COLOR_BGR2RGB)

    axes[i].imshow(imagen_color)
    axes[i].set_title(f'{valor} - {nombre}', loc='left', fontsize=10, fontweight='bold')
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
